# 🎓 Batch 2028 B.Tech Software & Tech Roles Internship Intelligence Tracker
### 🚀 365+ Top Tech Companies • Verified Careers Portals • Stipend Breakdown • Selection Insights

This notebook provides a complete interactive intelligence tracker for **Batch 2028 B.Tech / B.E. undergraduates** seeking software engineering, AI/ML, Data Science, Systems, DevOps, and Quant tech internships in 2026/2027/2028.


```
import sys
import os
import pandas as pd

# Add project root to path
sys.path.insert(0, os.path.abspath(".."))

from scrapers.internship_scraper import InternshipScraper
from scrapers.core.storage import DataStorage

scraper = InternshipScraper()
companies = scraper.get_all_companies()

# Convert to DataFrame
df = pd.DataFrame(companies)
df["roles_count"] = df["roles_offered"].apply(len)
df["tech_stack_str"] = df["tech_stack"].apply(lambda t: ", ".join(t))
df["locations_str"] = df["locations"].apply(lambda l: ", ".join(l))

print(f"✅ Total Hiring Companies Loaded: {len(df)}")
print(f"📊 Industry Sectors: {df["category"].nunique()}")
```

## 1. 📊 Sector & Category Distribution

```
cat_counts = df["category"].value_counts().reset_index()
cat_counts.columns = ["Industry Sector", "Total Companies"]
display(cat_counts)
```

## 2. 💎 Tier-1 High Frequency Trading (HFT) & FAANG Tech Internships

```
hft_faang = df[df["category"].str.contains("FAANG|HFT", case=False, regex=True)]
display(hft_faang[["company_name", "category", "stipend_range", "tech_stack_str", "careers_url", "interview_focus"]].head(15))
```

## 3. 🦄 Indian Unicorns & High-Growth Product Startups

```
unicorns = df[df["category"].str.contains("Unicorn", case=False)]
display(unicorns[["company_name", "stipend_range", "tech_stack_str", "locations_str", "careers_url"]].head(15))
```

## 4. 🧠 Generative AI, DeepTech & LLM Inference Labs

```
ai_companies = df[df["category"].str.contains("AI", case=False)]
display(ai_companies[["company_name", "stipend_range", "tech_stack_str", "careers_url", "interview_focus"]].head(15))
```

## 5. 🔍 Interactive Search by Tech Stack or Location

```
def search_companies(tech_skill="Python", location="Bangalore"):
    matched = df[
        df["tech_stack_str"].str.contains(tech_skill, case=False, na=False) &
        df["locations_str"].str.contains(location, case=False, na=False)
    ]
    print(f"Found {len(matched)} companies hiring for skill '{tech_skill}' in '{location}':")
    return matched[["company_name", "category", "stipend_range", "careers_url", "tech_stack_str"]]

search_companies(tech_skill="C++", location="Bangalore").head(10)
```
