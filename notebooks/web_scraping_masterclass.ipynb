# 🌐 Universal Web Scraping Intelligence Masterclass
A complete, interactive walkthrough covering 5 production-grade web scraping domains:
1. **Companies & Tech Stack Intelligence**
2. **Restaurants, Menus & Culinary Data**
3. **Video Games & Steam Platform Insights**
4. **Wikipedia Deep Infobox & Section Extraction**
5. **E-Commerce Product Catalogs & Price Monitoring**


```
import sys
import os
import pandas as pd

# Add project root to sys.path
sys.path.insert(0, os.path.abspath(".."))

from scrapers import (
    CompanyScraper,
    RestaurantScraper,
    GameScraper,
    WikipediaScraper,
    EcommerceScraper,
)
from scrapers.core.storage import DataStorage

storage = DataStorage(output_dir="../data/output")
print("✅ Web Scraping Toolkit initialized successfully!")
```

## 1. 🏢 Company & Tech Stack Scraper
Extracts corporate metadata, OpenGraph signals, tech tags, GitHub repositories, and open job roles.


```
c_scraper = CompanyScraper()
dossier = c_scraper.build_company_dossier("Stripe")

print(f"Company: {dossier["name"]} ({dossier["domain"]})")
print(f"HQ: {dossier["headquarters"]}")
print(f"Detected Tech Stack: {dossier["tech_stack"][:10]}")

# View jobs as a DataFrame
if dossier["recent_job_openings"]:
    df_jobs = pd.DataFrame(dossier["recent_job_openings"])
    display(df_jobs[["position", "location", "salary_range"]])
```

## 2. 🍽️ Restaurant & Culinary Intelligence Scraper
Scrapes restaurant listings, ratings, price tiers, operating hours, and structured menu items.


```
r_scraper = RestaurantScraper()
restaurant = r_scraper.scrape_restaurant_details("Bella Trattoria", city="New York, NY")

print(f"Restaurant: {restaurant["name"]}")
print(f"Rating: {restaurant["rating"]} ★ | Price: {restaurant["price_tier"]}")
print(f"Dietary: {restaurant["dietary_options"]}")

# Convert menu to DataFrame
df_menu = pd.DataFrame(restaurant["menu_items"])
display(df_menu[["category", "name", "price", "dietary"]])
```

## 3. 🎮 Video Games & Steam Intelligence Scraper
Fetches Steam store intelligence, hardware requirements, Metacritic ratings, and Free-to-play listings.


```
g_scraper = GameScraper()
game = g_scraper.scrape_game_details("1091500") # Cyberpunk 2077

print(f"Title: {game["title"]}")
print(f"Metacritic Score: {game.get("metacritic_score")}/100")
print(f"Developers: {game.get("developers")}")
print(f"Platforms: {game.get("platforms")}")
print(f"Minimum Requirements: {game.get("system_requirements", {}).get("minimum")[:150]}...")
```

## 4. 📖 Wikipedia Deep-Knowledge Scraper
Extracts structured infoboxes, hierarchical outline sections, tables into DataFrames, and references.


```
w_scraper = WikipediaScraper(language="en")
wiki_article = w_scraper.scrape_article("Artificial_intelligence")

print(f"Article: {wiki_article["title"]}")
print(f"Summary: {wiki_article["summary"][:300]}...
")
print(f"Total Sections: {len(wiki_article["sections"])}")
print(f"Infobox Attributes: {list(wiki_article["infobox"].keys())[:6]}")
```

## 5. 🛍️ E-Commerce & Product Catalog Scraper
Extracts product catalogs, live prices, star ratings, UPC codes, and availability.


```
e_scraper = EcommerceScraper()
products = e_scraper.scrape_category_products("Science Fiction", max_pages=1)

df_products = pd.DataFrame(products)
df_products["price_amount"] = df_products["price"].apply(lambda p: p.get("amount") if isinstance(p, dict) else 0.0)
display(df_products[["title", "price_amount", "rating", "availability"]].head(10))
```

## 6. 🗄️ Querying Scraped Data via SQLite Database


```
# Query stored records from SQLite
for table in ["companies", "restaurants", "games", "wikipedia_articles", "ecommerce_products"]:
    df_table = storage.query_db(f"SELECT * FROM {table} LIMIT 5")
    print(f"=== Table: {table} ({len(df_table)} rows preview) ===")
    display(df_table.head(2))
```
